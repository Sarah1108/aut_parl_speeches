```python
#Function to visualise the seats in parliament 

def plot_seats(allocations,parties,party_colors,title):

    ax = poli_sci_kit.plot.parliament(
        allocations=allocations,
        labels=parties,
        colors=party_colors,
        style="semicircle", # "rectangle"
        num_rows=4,
        marker_size=183,
        speaker=False,
        df_seat_lctns=None,
        axis=None,
    )

    handles, labels = pltviz.legend.gen_elements(
        counts=allocations,
        labels=parties,
        colors=party_colors,
        size=8,
        marker="o",
        padding_indexes=None,
        order=[0,1,2,3,4],
    )
    ax.legend(
        handles=handles,
        labels=labels,
        title="Seats per Party: {} total".format(sum(allocations)),
        title_fontsize=12,
        fontsize=10,
        ncol=2,
        loc="center",
        bbox_to_anchor=(0.5, 0.12),
        frameon=False,
        facecolor="#ffffff",
        framealpha=1,
        
    )
    
    ax.yaxis.grid(color='gray', linestyle='dashed')
    
    plt.title(title)



    plt.show()
```

```python
def plot_speech_seats_dispr(allocations, speeches, parties, party_colors, title):
    ax = poli_sci_kit.plot.dispr_bar(
        shares=speeches,
        allocations=allocations,
        labels=parties,
        colors=None,
        total_shares=sum(speeches),
        total_alloc=sum(allocations),
        percent=True,
        axis=None,
    )

    for bar, color in zip(ax.patches, party_colors):
        bar.set_facecolor(color)
        
    handles, labels = pltviz.legend.gen_elements(
        counts=[round(v / sum(speeches) * 100, 1) for v in speeches],
        labels=parties,
        colors=party_colors,
        size=11,
        marker="o",
        padding_indexes=None,
        order=None,
    )

    ax.legend(
        handles=handles,
        labels=labels,
        title="Speech Percents",
        title_fontsize=10,
        fontsize=8,
        ncol=1,
        loc="best",#best
        bbox_to_anchor=(0, 1),
        frameon=True,
        facecolor="#ffffff",
        framealpha=1,
    )

    ax.set_title(title, fontsize=12)
    ax.set_xlabel("Parties", fontsize=10)
    ax.set_ylabel("Percent Shift", fontsize=10)
    ax.yaxis.grid(color='gray', linestyle='dashed')
    ax.set_axisbelow(True)

    #for bar, value in zip(ax.patches, [round(a - s, 2) for a, s in zip(allocations, speeches)]):
    #    height = bar.get_height()
    #    ax.annotate(
    #        f"{value:+.2f}",
    #        xy=(bar.get_x() + bar.get_width() / 2, height),
    #        xytext=(0, 5 if height >= 0 else -15),  # shift text slightly above/below bar
    #        textcoords="offset points",
    #        ha='center',
    #        va='bottom' if height >= 0 else 'top',
    #        fontsize=11,
    #        color='black'
    #    )

    plt.show()
```

```python
def assign_gov(date):
    if date >= pd.Timestamp('1996-01-15') and date < pd.Timestamp('1999-10-29'):
        return 0
    elif date >= pd.Timestamp('1999-10-29') and date < pd.Timestamp('2002-12-20'):
        return 1
    elif date >= pd.Timestamp('2002-12-20') and date < pd.Timestamp('2006-10-30'):
        return 2
    elif date >= pd.Timestamp('2006-10-30') and date < pd.Timestamp('2008-10-28'):
        return 3 
    elif date >= pd.Timestamp('2008-10-28') and date < pd.Timestamp('2013-10-29'):
        return 4
    elif date >= pd.Timestamp('2013-10-29') and date < pd.Timestamp('2017-11-08'):
        return 5 
    elif date >= pd.Timestamp('2017-11-08') and date < pd.Timestamp('2019-10-22'):
        return 6
    else:
        return 7
```

```python
#load data frame with all speeches
fn="data/Arbeitskopie_BERT_inkl_indie.xlsx"
df= pd.read_excel(fn, index_col=0)
```

```python
df['date']= pd.to_datetime(df['date'])
df['gov'] = df['date'].map(assign_gov)
df= df.sort_values('date')
df = df.set_index('date')
```

```python
df['speaker'] = np.where((df['speaker'] == 'Hans Moser') & (df['gov'] == 2),'Johann Moser',df['speaker'])
#evtl noch was bereinigen
```

```python
df['speaker'] = df['speaker'].replace(to_replace="Beate Hartinger$", value="Beate Hartinger-Klein", regex=True)
df['speaker'] = df['speaker'].replace(to_replace="Karin Miklautsch$", value= "Karin Gastinger", regex=True)
df['speaker'] = df['speaker'].replace(to_replace= "Helmut Moser$", value= "Hans Moser",regex=True)
df['speaker'] = df['speaker'].replace(to_replace="Michael Pock$", value="Michael Bernhard", regex=True)
df['speaker'] = df['speaker'].replace(to_replace="Elisabeth Aumayr$", value= "Elisabeth Achatz", regex=True)
df['speaker'] = df['speaker'].replace(to_replace="Carmen Gartelgruber$", value= "Carmen Schimanek", regex=True)
df['speaker'] = df['speaker'].replace(to_replace="Sylvia Prettenthaler$", value= "Sylvia Rinner", regex=True)
```

```python
# set all party labels, colors and number of speeches per party
all_parties= ['SPÖ','ÖVP','FPÖ','Grüne','BZÖ','LIF','NEOS','independent','STRONACH','PILZ']
all_party_colors= ['#ff0000','#000000','#0066ff','#92d050','#ee7f00','#fecd00','#e84188','#a9a9a9','#ffff40','#efebdc']
sns.palplot(sns.color_palette(all_party_colors))
```
